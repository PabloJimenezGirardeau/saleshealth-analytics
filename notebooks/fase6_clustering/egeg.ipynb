```python
from sqlalchemy import create_engine, text
import pandas as pd

engine = create_engine('postgresql+psycopg2://postgres:pimpum.postgre@localhost:5432/saleshealth')

def query(sql):
    with engine.connect() as conn:
        return pd.read_sql_query(text(sql), conn)

result = query("""
    SELECT cluster_label, COUNT(*) AS n,
           ROUND(AVG(cltv)::numeric, 2) AS cltv_medio,
           ROUND(MIN(cltv)::numeric, 2) AS cltv_min,
           ROUND(MAX(cltv)::numeric, 2) AS cltv_max
    FROM marts.customer_360
    WHERE cltv IS NOT NULL
    GROUP BY cluster_label
    ORDER BY cltv_medio DESC
""")
print(result.to_string(index=False))
```
